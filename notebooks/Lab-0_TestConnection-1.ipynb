# Lab-0. Connect to spark claster without Delta Lake support

```python
from pyspark.sql import SparkSession

spark = SparkSession.builder \
    .master("spark://spark-master:7077") \
    .appName("TestSparkConnection") \
    .getOrCreate()

print(f"Spark version: {spark.version} without Delta support")
```

## Run this task to check parallel calcilation

```python
data = range(1, 1000000)
distData = spark.sparkContext.parallelize(data, 10) # розбиваємо на 10 частин

result = distData.sum()
print(f"Сума: {result}")
```
